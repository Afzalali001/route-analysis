```python
import pandas as pd

df = pd.read_csv("../data/processed/clean_shipping_data.csv")
```

```python
product_factory = {
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar -Scrumdiddlyumptious": "Lot's O' Nuts",
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",
    "Everlasting Gobstopper": "Secret Factory",
    "Hair Toffee": "The Other Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",
    "Kazookles": "The Other Factory"
}
```

```python
df["Factory"] = df["Product Name"].map(product_factory)
```

```python
df[["Product Name", "Factory"]].head(15)
```

```python
df["Factory"].isnull().sum()
```

```python
factory_coordinates = {
    "Lot's O' Nuts": (32.881893, -111.768036),
    "Wicked Choccy's": (32.076176, -81.088371),
    "Sugar Shack": (48.119140, -96.181150),
    "Secret Factory": (41.446333, -90.565487),
    "The Other Factory": (35.117500, -89.971107)
}
```

```python
df["Factory Latitude"] = df["Factory"].map(
    lambda x: factory_coordinates[x][0]
)
```

```python
df["Factory Longitude"] = df["Factory"].map(
    lambda x: factory_coordinates[x][1]
)
```

```python
df[[
    "Factory",
    "Factory Latitude",
    "Factory Longitude"
]].head(10)
```

```python
df["Factory_State_Route"] = (
    df["Factory"] + " → " + df["State/Province"]
)
```

```python
df[["Factory", "State/Province", "Factory_State_Route"]].head(10)
```

```python
df["Factory_Region_Route"] = (
    df["Factory"] + " → " + df["Region"]
)
```

```python
df[["Factory", "Region", "Factory_Region_Route"]].head(10)
```

```python
df.columns
```

```python
region_summary = (
    df.groupby("Region")
      .agg(
          Total_Shipments=("Order ID", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean")
      )
      .reset_index()
)
```

```python
region_summary
```

```python
state_summary = (
    df.groupby("State/Province")
      .agg(
          Total_Shipments=("Order ID", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean")
      )
      .reset_index()
)
```

```python
state_summary.head()
```

```python
#top ten bottleneck state
bottleneck_states = state_summary.sort_values(
    by="Average_Lead_Time",
    ascending=False
)

bottleneck_states.head(10)
```

```python
#state with high volum but low lead time
high_volume_poor_performance = state_summary[
    (state_summary["Total_Shipments"] > state_summary["Total_Shipments"].median()) &
    (state_summary["Average_Lead_Time"] > state_summary["Average_Lead_Time"].median())
]

high_volume_poor_performance
```

```python
region_summary.to_csv(
    "../data/processed/region_summary.csv",
    index=False
)

state_summary.to_csv(
    "../data/processed/state_summary.csv",
    index=False
)

bottleneck_states.to_csv(
    "../data/processed/bottleneck_states.csv",
    index=False
)
```
