```python
import pandas as pd

df_clean = pd.read_csv("../data/processed/feature_engineered_data.csv")
```

```python
df_clean.head()
```

```python
# info or summary of factory by state
route_state = (
    df_clean.groupby("Factory_State_Route")
      .agg(
          Total_Shipments=("Order ID", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean"),
          Lead_Time_Variability=("Shipping Lead Time", "std")
      )
      .reset_index()
)
```

```python
route_state.head(10)
```

```python
# save summary
route_state.to_csv(
    "../data/processed/route_state_summary.csv",
    index=False
)
```

```python
route_summary = pd.read_csv("../data/processed/route_state_summary.csv")
```

```python
route_summary.head()
```

```python
route_summary = route_summary.sort_values(
    by="Average_Lead_Time",
    ascending=True
)
```

```python
route_summary["Route_Rank"] = range(
    1,
    len(route_summary) + 1
)
```

```python
top10_routes = route_summary.head(10)

top10_routes
```

```python
bottom10_routes = route_summary.tail(10)

bottom10_routes
```

```python
top10_routes.to_csv(
    "../data/processed/top_10_routes.csv",
    index=False
)

bottom10_routes.to_csv(
    "../data/processed/bottom_10_routes.csv",
    index=False
)
```

```python
route_summary = df_clean.groupby("Factory_State_Route").agg(
    Average_Lead_Time=("Shipping Lead Time", "mean"),
    Route_Volume=("Factory_State_Route", "count")
).reset_index()
```

```python
route_summary.head()
```

```python
route_summary.to_csv(
    "../data/processed/route_state_summary.csv",
    index=False
)
```

```python
pd.read_csv("../data/processed/route_state_summary.csv").head()
```

```python
max_time = route_summary["Average_Lead_Time"].max()
min_time = route_summary["Average_Lead_Time"].min()

route_summary["Route_Efficiency_Score"] = (
    100 * (max_time - route_summary["Average_Lead_Time"])
    / (max_time - min_time)
)
```

```python
route_summary.head()
```

```python
route_summary.to_csv(
    "../data/processed/route_state_summary.csv",
    index=False
)
```
