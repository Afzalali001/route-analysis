```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/feature_engineered_data.csv")
```

```python
ship_mode_summary = (
    df.groupby("Ship Mode")
      .agg(
          Total_Shipments=("Order ID", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean"),
          Average_Sales=("Sales", "mean"),
          Average_Cost=("Cost", "mean")
      )
      .reset_index()
)
```

```python
ship_mode_summary
```

```python
delay_summary = (
    df.groupby("Ship Mode")["Delay Flag"]
      .apply(lambda x: (x == "Delayed").mean() * 100)
      .reset_index(name="Delay_Percentage")
)
```

```python
ship_mode_summary = ship_mode_summary.merge(
    delay_summary,
    on="Ship Mode"
)
```

```python
ship_mode_summary
```

```python
fastest_mode = ship_mode_summary.sort_values(
    by="Average_Lead_Time"
)

fastest_mode.head()
```

```python
highest_delay = ship_mode_summary.sort_values(
    by="Delay_Percentage",
    ascending=False
)

highest_delay.head()
```

```python
ship_mode_summary.to_csv(
    "../data/processed/ship_mode_summary.csv",
    index=False
)
```
